```python
class Animal:
    def __init__(self,name,species):
        self.name=name
        self.species=species
class dog(Animal):
    def __init__(self,name,species,breed):
        super().__init__(name,species)
        self.breed=breed
    def bark(self):
        return f"{self.name} bark woff"
    def __str__(self):
        return f'name:{self.name},species:{self.species},breed:{self.breed}'

dog=dog("rocky","animal","labrado")
print(dog)
print(dog.bark())

```

```python
class A:
    def show(self):
        print("showing")
class B(A):
    def show(self):
        print("showing B")
class C(A):
    def show(self):
        print("latest show") 
class D(B,C):
    pass

c=D()
c.show()
```

```python
class Shape:
    def __init__(self,colour):
        self.colour=colour
class Circle(Shape):
    def __init__(self,colour,radius):
        super().__init__(colour)
        self.radius=radius
c=Circle("blue",12)
print(c.colour,c.radius)
```

```python
class Vehicle:
    def start(self):
        print("the vehicle is starting")
class Car(Vehicle):
    def start(self):
        print("the car is starting")
        super().start()
car=Car()
car.start() 
```

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee:
    def __init__(self, employee_id):
        self.employee_id = employee_id

class Manager(Person, Employee):
    def __init__(self, name, employee_id):
        super().__init__(name)
        Employee.__init__(self, employee_id)
p=Manager("kaiku",2)
print(p.name,p.employee_id)
```

```python
class Base1:
    def __init__(self,a):
        self.a=a

class Base2:
    def __init__(self,b):
        self.b=b
class Derived(Base1,Base2):
    def __init__(self,a,b,c):
        super().__init__(a)
        Base2.__init__(self,b)
        self.c=c
d=Derived(1,2,3)
print(d.a,d.b,d.c)
 
 
```

```python
class Device:
    def __init__(self,brand):
        self.brand=brand
class Camera:
    def __init__(self,resolution):
        self.resolution=resolution
class Phone(Device):
    def __init__(self,brand,model):
        super().__init__(brand)
        self.model=model
class Smartphone(Phone,Camera):
    def __init__(self,brand,model,resolution):
        super().__init__(brand,model)
        Camera.__init__(self,resolution)

phone=Smartphone("nokia",2023,"8k")
print(phone.brand,phone.model,phone.resolution)

    

```
