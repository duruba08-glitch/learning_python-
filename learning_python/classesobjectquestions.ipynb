```python
class vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"


a=vector(2,5)
b=vector(7,6)
c=a+b
print(c)
       

    
    
```

```python
class contextmanager:
    def __init__(self,filename,mode):
        self.filename=filename
        self.mode= mode
    def __enter__(self):
        self.file= open( "self.filename","self.mode")
        return self.file
    def __exit__(self,exc_type,exc_value,traceback):
        self.file.close()

with contextmanager('sample.txt', 'r') as file:
      content = file.read()
      print(content) 
        
```

```python
class calculator:
    def __init__(self,x=0):
        self.x=x
    def add(self,amount):
         self.x+=amount
         return self
    def sub(self,amount):
        self.x-=amount
        return self
    def mul(self,amount):
        self.x*=amount
        return self
    def div(self,amount):
        if amount != 0:
            self.x /= amount
        else:
            print("Cannot divide by zero!")
        return self

a=calculator()
a.add(1).sub(2).mul(3).div(4)
print(a.x)
```
