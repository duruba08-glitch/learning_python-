```python
def replace(path,word):
    with open(path,'r') as f:
        content= f.read()
    content=content.replace(word,"kaiku")
    return content
print(replace("simple.txt", "i"))
      
```

```python
def reverse(path):
    with open(path,'r') as f:
        content=f.readlines()
    for i in reversed(content):
        print(i.strip()) 

    

reverse("simple.txt") 
```

```python
def count(path):
    with open(path, 'r') as f:
        content = f.read()

    lines = content.splitlines()
    words = content.split()
    chars = len(content)

    print(len(lines))   # lines
    print(len(words))   # words
    print(chars)        # characters
       
       
count("simple.txt")
```
